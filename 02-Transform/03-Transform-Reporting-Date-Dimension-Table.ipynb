```python
import pandas as pd
```

```python
source_date_pd_df = pd.date_range(start = '2023-05-01', end = '2025-12-31')
source_date_spark_df = spark.createDataFrame(source_date_pd_df.to_frame())
```

```python
source_date_spark_df.withColumnRenamed('0', 'calendar_date').createOrReplaceTempView('source_date_dim')
```

```python
%sql
select * from source_date_dim
```
